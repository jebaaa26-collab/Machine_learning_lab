```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

data = {
    "Week": [1, 2, 3, 4, 5],
    "Sales": [1.2, 1.8, 2.6, 3.2, 3.8]
}

df = pd.DataFrame(data)

print("========== DATASET ==========")
print(df)

X = df[["Week"]]
Y = df["Sales"]


model = LinearRegression()
model.fit(X, Y)

predictions = model.predict(X)

print("\n========== MODEL ==========")
print("Intercept :", model.intercept_)
print("Coefficient :", model.coef_[0])

print("\nRegression Equation")
print(f"Sales = {model.coef_[0]:.2f} × Week + {model.intercept_:.2f}")


print("\n========== PREDICTIONS ==========")

for week, actual, predicted in zip(df["Week"], Y, predictions):
    print(f"Week {week} | Actual = {actual:.2f} | Predicted = {predicted:.2f}")


mse = mean_squared_error(Y, predictions)
mae = mean_absolute_error(Y, predictions)
rmse = np.sqrt(mse)
r2 = r2_score(Y, predictions)

print("\n========== LOSS FUNCTIONS ==========")
print(f"Mean Squared Error (MSE) : {mse:.4f}")
print(f"Mean Absolute Error (MAE): {mae:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")

print("\n========== MODEL SCORE ==========")
print(f"R² Score : {r2:.4f}")


plt.figure(figsize=(7,5))

plt.scatter(X, Y,
            color="green",
            s=80,
            label="Actual Data")


plt.plot(X,
         predictions,
         color="blue",
         linewidth=2,
         label="Regression Line")


plt.title("Linear Regression: Week vs Sales")
plt.xlabel("Week")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)

plt.show()
```
