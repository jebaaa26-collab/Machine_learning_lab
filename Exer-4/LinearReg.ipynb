```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression


data = {
    "Week": [1, 2, 3, 4, 5],
    "Sales": [1.2, 1.8, 2.6, 3.2, 3.8]
}


df = pd.DataFrame(data)


print("Dataset:")
print(df)


X = df[["Week"]]      # Independent variable
Y = df["Sales"]       # Dependent variable


model = LinearRegression()
model.fit(X, Y)


predictions = model.predict(X)


print("\nIntercept:", model.intercept_)
print("Coefficient:", model.coef_[0])


print("\nRegression Equation:")
print(f"Sales = {model.coef_[0]:.2f} * Week + {model.intercept_:.2f}")


print("\nPredicted Sales:")
for week, actual, predicted in zip(df["Week"], Y, predictions):
    print(f"Week {week}: Actual = {actual}, Predicted = {predicted:.2f}")


print(f"\nR² Score: {model.score(X, Y):.4f}")


plt.figure(figsize=(6, 4))
plt.scatter(X, Y, color="green", label="Actual Data")


plt.plot(X, predictions, color="blue", linewidth=2, label="Regression Line")


plt.title("Linear Regression: Week vs Sales")
plt.xlabel("Week")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)


plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.datasets import make_regression

# Generate a random regression dataset
X, y = make_regression(
    n_samples=50,
    n_features=1,
    noise=0.1,
    random_state=42
)

# Plot dataset
plt.figure(figsize=(6, 4))
plt.scatter(X, y, color="green", label="Data Points")

# Create and train model
model = LinearRegression()
model.fit(X, y)

# Predict values
predictions = model.predict(X)

# Print model details
print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])

print("\nRegression Equation:")
print(f"y = {model.coef_[0]:.2f} * X + {model.intercept_:.2f}")

print(f"\nR² Score: {model.score(X, y):.4f}")

# Plot regression line
plt.plot(X, predictions, color="red", linewidth=2, label="Regression Line")

plt.title("Simple Linear Regression")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.datasets import make_regression

print("Multiple Linear Regression\n")

# Generate dataset
X, y = make_regression(
    n_samples=50,
    n_features=2,
    noise=0.1,
    random_state=42
)

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Print intercept
print("Intercept:", model.intercept_)

# Print coefficients
print("\nCoefficients:")
for i, coef in enumerate(model.coef_, start=1):
    print(f"Coefficient of x{i}: {coef:.4f}")

# Regression equation
print("\nRegression Equation:")
print(
    f"y = {model.coef_[0]:.4f} * x1 + "
    f"{model.coef_[1]:.4f} * x2 + "
    f"{model.intercept_:.4f}"
)

# R² Score
print(f"\nR² Score: {model.score(X, y):.4f}")
```

```python
import numpy as np
from sklearn.decomposition import PCA

# Original dataset
X = np.array([
    [2, 1],
    [3, 2],
    [4, 3],
    [5, 4],
    [6, 5]
])

print("Original Matrix:")
print(X)
print("Shape:", X.shape)

# Create PCA model
pca = PCA(n_components=2)

# Apply PCA
X_pca = pca.fit_transform(X)

print("\nTransformed Matrix:")
print(X_pca)
print("Shape:", X_pca.shape)

# Reconstruct original data
X_reconstructed = pca.inverse_transform(X_pca)

print("\nReconstructed Matrix:")
print(X_reconstructed)

# Explained variance ratio
print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

# Singular values
print("\nSingular Values:")
print(pca.singular_values_)

# Principal components
print("\nPrincipal Components:")
print(pca.components_)

# Mean of dataset
print("\nMean of Dataset:")
print(pca.mean_)
```
