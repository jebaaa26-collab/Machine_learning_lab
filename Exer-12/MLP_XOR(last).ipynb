```python
import numpy as np
class NeuralNetwork:

    def __init__(self):

        self.w13 = -0.2
        self.w23 = 0.2

        self.w14 = 0.4
        self.w24 = -0.3
     
        self.w35 = 0.2
        self.w45 = -0.3
      
        self.theta3 = -0.4
        self.theta4 = 0.1
        self.theta5 = -0.3
   
        self.learning_rate = 0.8

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def sigmoid_derivative(self, x):
        return x * (1 - x)

    def forward(self, x1, x2):

        net3 = (
            x1 * self.w13 +
            x2 * self.w23 +
            self.theta3
        )

        x3 = self.sigmoid(net3)

        net4 = (
            x1 * self.w14 +
            x2 * self.w24 +
            self.theta4
        )

        x4 = self.sigmoid(net4)

        net5 = (
            x3 * self.w35 +
            x4 * self.w45 +
            self.theta5
        )
        x5 = self.sigmoid(net5)
        return x3, x4, x5
    def train(self, X, Y, epochs=10000):
        for epoch in range(epochs):
            total_error = 0
            for i in range(len(X)):
                x1 = X[i][0]
                x2 = X[i][1]
                target = Y[i][0]
                x3, x4, x5 = self.forward(x1, x2)
                error = target - x5
                total_error += error ** 2
                delta5 = (
                    error *
                    self.sigmoid_derivative(x5)
                )
                delta3 = (
                    self.sigmoid_derivative(x3)
                    * self.w35
                    * delta5
                )
                delta4 = (
                    self.sigmoid_derivative(x4)
                    * self.w45
                    * delta5
                )
                self.w35 += (
                    self.learning_rate *
                    delta5 *
                    x3
                )

                self.w45 += (
                    self.learning_rate *
                    delta5 *
                    x4
                )

                self.w13 += (
                    self.learning_rate *
                    delta3 *
                    x1
                )
                self.w23 += (
                    self.learning_rate *
                    delta3 *
                    x2
                )

                self.w14 += (
                    self.learning_rate *
                    delta4 *
                    x1
                )
                self.w24 += (
                    self.learning_rate *
                    delta4 *
                    x2
                )

                self.theta5 += (
                    self.learning_rate * delta5
                )

                self.theta3 += (
                    self.learning_rate * delta3
                )

                self.theta4 += (
                    self.learning_rate * delta4
                )


            if epoch % 1000 == 0:

                print(
                    "Epoch:",
                    epoch,
                    "Error:",
                    round(total_error, 6)
                )


    def predict(self, X):

        predictions = []
        for row in X:

            x1 = row[0]
            x2 = row[1]

            x3, x4, x5 = self.forward(x1, x2)

            prediction = 1 if x5 >= 0.5 else 0

            predictions.append(
                (int(x1), int(x2), x5, prediction)
            )
        return predictions
```

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=float)

Y = np.array([
    [0],
    [1],
    [1],
    [0]
], dtype=float)

print("Input:")
print(X)

print("\nExpected Output:")
print(Y)
```
