```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-np.pi,np.pi,100)

y = np.sin(x)

plt.plot(x,y)

plt.title("Sine Wave")
plt.xlabel("Angle")
plt.ylabel("sin(x)")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0,10,100)

plt.subplot(2,1,1)
plt.plot(x,np.sin(x))
plt.title("Sine")

plt.subplot(2,1,2)
plt.plot(x,np.cos(x))
plt.title("Cosine")

plt.show()
```

```python
tips = sns.load_dataset("tips")
print(tips)
```

```python
print(tips.head())
```

```python
sns.pairplot(tips)
plt.show()
```

```python
sns.histplot(tips["total_bill"])

plt.show()
```

```python
sns.scatterplot(
    x="total_bill",
    y="tip",
    data=tips
)

plt.show()
```

```python
sns.countplot(
    x="day",
    data=tips
)

plt.show()
```

```python
sns.boxplot(
    x="day",
    y="total_bill",
    data=tips
)

plt.show()
```

```python
corr = tips.corr(numeric_only=True)

sns.heatmap(corr,
            annot=True)

plt.show()
```

```python
import seaborn as sns

import matplotlib.pyplot as plt



tips = sns.load_dataset("tips")



print(tips.head())



sns.histplot(tips["total_bill"])

plt.show()



sns.scatterplot(x="total_bill",

                y="tip",

                data=tips)

plt.show()



sns.boxplot(x="day",

            y="total_bill",

            data=tips)

plt.show()



sns.countplot(x="day",

              data=tips)

plt.show()



corr = tips.corr(numeric_only=True)



sns.heatmap(corr,

            annot=True)



plt.show()
```
