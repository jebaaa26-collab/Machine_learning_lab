```python

states = ('Good', 'Moderate', 'Fail')

initial = {'Good':0.5, 'Moderate':0.25, 'Fail':0.25}

transition = {
    'Good': {'Good':0.4, 'Moderate':0.3, 'Fail':0.3},
    'Moderate': {'Good':0.6, 'Moderate':0.25, 'Fail':0.15},
    'Fail': {'Good':0.3, 'Moderate':0.4, 'Fail':0.3}
}

emission = {
    'Good': {'Happy':0.4, 'Worry':0.1, 'Blink':0.2, 'Fine':0.3},
    'Moderate': {'Happy':0.3, 'Worry':0.2, 'Blink':0.3, 'Fine':0.2},
    'Fail': {'Happy':0.1, 'Worry':0.4, 'Blink':0.3, 'Fine':0.2}
}

def viterbi(obs):
    prob = {s: initial[s] * emission[s][obs[0]] for s in states}
    paths = {s: [s] for s in states}

    for o in obs[1:]:
        new_prob = {}
        new_paths = {}
        for s in states:
            p, prev = max(
                (prob[x] * transition[x][s] * emission[s][o], x)
                for x in states
            )
            new_prob[s] = p
            new_paths[s] = paths[prev] + [s]
        prob, paths = new_prob, new_paths

    return paths[max(states, key=lambda s: prob[s])]

obs = ['Happy', 'Blink', 'Fine', 'Fine']
print("Observations:", obs)
print("Hidden states:", viterbi(obs))
```
