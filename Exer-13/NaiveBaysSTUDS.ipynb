```python
import pandas as pd

data = {
    'CGPA': ['>=9', '>=8', '>=9', '<8', '>=8',
             '>=9', '<8', '>=9', '>=8', '>=8'],

    'Interactiveness': ['Yes', 'No', 'No', 'No', 'Yes',
                        'Yes', 'Yes', 'No', 'Yes', 'Yes'],

    'Practical_Knowledge': ['Very Good', 'Good', 'Average', 'Average',
                            'Good', 'Good', 'Good', 'Very Good',
                            'Good', 'Average'],

    'Communication_Skills': ['Good', 'Moderate', 'Poor', 'Good',
                             'Moderate', 'Moderate', 'Poor', 'Good',
                             'Good', 'Good'],

    'Job_Offer': ['Yes', 'Yes', 'No', 'No', 'Yes',
                  'Yes', 'No', 'Yes', 'Yes', 'Yes']
}

df = pd.DataFrame(data)

print(df)
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.naive_bayes import CategoricalNB

# Convert categorical values into numbers
encoders = {}

for column in df.columns:
    encoders[column] = LabelEncoder()
    df[column] = encoders[column].fit_transform(df[column])

# Separate input and output
X = df.drop('Job_Offer', axis=1)
y = df['Job_Offer']

# Create Naive Bayes classifier
model = CategoricalNB()

# Train the model
model.fit(X, y)

# Test student
test_student = pd.DataFrame({
    'CGPA': ['>=9'],
    'Interactiveness': ['Yes'],
    'Practical_Knowledge': ['Average'],
    'Communication_Skills': ['Good']
})

# Encode test data using the same encoders
for column in test_student.columns:
    test_student[column] = encoders[column].transform(test_student[column])

# Prediction
prediction = model.predict(test_student)

# Convert prediction back to Yes/No
result = encoders['Job_Offer'].inverse_transform(prediction)

print("Job Offer Prediction:", result[0])
```

```python
probability = model.predict_proba(test_student)

print("Probability of No Job Offer :", probability[0][0])
print("Probability of Job Offer    :", probability[0][1])
```
