```python
import pandas as pd
from matplotlib import pyplot as plt

col_list = ['id','name','gender','physics','chemistry','maths','biology','language','selected']

df = pd.read_csv('marks1.csv', usecols=col_list)

print(df)
```

```python
marks1 = df['physics'].values.tolist()

df['physics'].hist()

plt.title("Histogram for Physics Marks")
plt.xlabel("Physics Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
chemistry1 = df['chemistry'].values.tolist()

plt.scatter(chemistry1, marks1, alpha=0.5)

plt.title("Chemistry vs Physics")
plt.xlabel("Chemistry")
plt.ylabel("Physics")

plt.show()
```

```python
df.plot.box(title="Box and Whisker Plot", grid=False)

plt.show()
```

```python
sums = df.selected.groupby(df.gender).sum()

plt.pie(sums, labels=sums.index, autopct="%1.1f%%")

plt.title("Selected Students by Gender")

plt.show()
```

```python
df.plot(kind='box',
        subplots=True,
        layout=(3,3),
        sharex=False,
        sharey=False)

plt.show()
```

```python
cols = df[['id','name','gender','selected']]

df2 = df.drop(cols, axis=1)

pd.plotting.scatter_matrix(df2, alpha=0.3)

plt.show()
```

```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=[
        'sepal.length',
        'sepal.width',
        'petal.length',
        'petal.width'
    ]
)

df['variety'] = [iris.target_names[i] for i in iris.target]

df.to_csv("iris.csv", index=False)

print(df.head())
```

```python
import pandas as pd

df = pd.read_csv("iris.csv")

data = df.iloc[1:20]

print(data)
```

```python
data.plot.scatter(
    x='sepal.length',
    y='sepal.width'
)

plt.show()
```

```python
data.hist(column='sepal.length')

plt.show()
```

```python
data.boxplot(
    column='sepal.length',
    by='variety'
)

plt.show()
```
