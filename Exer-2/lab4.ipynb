```python
import seaborn as sns
import pandas as pd
from matplotlib import pyplot as plt
```

```python
data = {
    'week':[1,2,3,4,5],
    'sales':[1.2,1.8,2.6,3.2,3.8],
    'region':['south','south','north','north','north']
}

df = pd.DataFrame(data)

print(df)
```

```python
df1 = sns.load_dataset("iris")

print(df1.head())
```

```python
sns.set(color_codes=True)
```

```python
sns.barplot(
    x="week",
    y="sales",
    hue="region",
    data=df
).set_title("Bar Plot for Week vs Sales")

plt.show()
```

```python
sns.boxplot(data=df)

plt.title("Box Plot")

plt.show()
```

```python
sns.histplot(df1["sepal_width"], kde=False)

plt.title("Distribution Plot")

plt.xlabel("Sepal Width")

plt.show()
```

```python
sns.kdeplot(df1["sepal_width"])

plt.title("Kernel Density Plot")

plt.show()
```

```python
sns.violinplot(data=df)

plt.title("Violin Plot")

plt.show()
```

```python
sns.countplot(
    x="region",
    data=df,
    palette="Blues"
)

plt.title("Count Plot")

plt.show()
```

```python
sns.stripplot(
    x="species",
    y="petal_length",
    data=df1
)

plt.title("Strip Plot")

plt.show()
```

```python
sns.jointplot(
    x="week",
    y="sales",
    data=df
)

plt.show()
```

```python
sns.lmplot(
    x="week",
    y="sales",
    hue="region",
    data=df
)
```

```python
sns.pairplot(
    df1,
    hue="species",
    kind="scatter"
)

plt.show()
```
