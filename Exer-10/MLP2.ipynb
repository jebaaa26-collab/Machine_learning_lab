```python
import numpy as np


class Perceptron:

    def __init__(self, input_size, lr=0.2, epochs=10):

        # Initial weights
        self.W = np.array([0.3, -0.2])

        self.lr = lr
        self.epochs = epochs

    # Step activation function
    def activation_fn(self, x):
        return 1 if x >= 0 else 0

    # Prediction
    def predict(self, x, theta):

        z = self.W.dot(x) - theta

        z = round(z, 2)

        return self.activation_fn(z)

    # Training
    def fit(self, X, d, theta):

        for epoch in range(self.epochs):

            print("\nEpoch:", epoch + 1)

            for i in range(len(X)):

                x = X[i]

                print("Input:", x,
                      "Weight:", self.W)

                # Prediction
                y = self.predict(x, theta)

                # Error
                e = d[i] - y

                print("Target:", d[i],
                      "Output:", y,
                      "Error:", e)

                # Weight update
                self.W = self.W + self.lr * e * x


# --------------------------------
# MAIN PROGRAM
# --------------------------------

if __name__ == "__main__":

    # AND gate inputs
    X = np.array([
        [0, 0],
        [0, 1],
        [1, 0],
        [1, 1]
    ])

    # AND gate expected output
    d = np.array([
        0,
        0,
        0,
        1
    ])

    # Create perceptron
    perceptron = Perceptron(
        input_size=2,
        lr=0.2,
        epochs=10
    )

    # Threshold
    theta = 0.4

    # Train
    perceptron.fit(X, d, theta)

    # Final weights
    print("\nFinal Weights:")
    print(perceptron.W)

    # Test the perceptron
    print("\nFinal Predictions:")

    for x in X:

        y = perceptron.predict(x, theta)

        print(x, "->", y)
```
