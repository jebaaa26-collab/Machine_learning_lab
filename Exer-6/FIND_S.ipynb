```python
import csv


def load_data(filename):
    with open(filename, "r") as file:
        data = list(csv.reader(file))
    return data



def display_data(data):
    print("Training Data:\n")
    for row in data:
        print(row)



def initialize_hypothesis(num_attributes):
    return ['Ø'] * num_attributes



def find_s_algorithm(data):

    # Number of attributes (excluding target column)
    num_attributes = len(data[0]) - 1

   
    hypothesis = initialize_hypothesis(num_attributes)

   
    for row in data[1:]:

      
        if row[-1].strip().lower() == "yes":

       
            for i in range(num_attributes):

                if hypothesis[i] == 'Ø':
                    hypothesis[i] = row[i]

                elif hypothesis[i] != row[i]:
                    hypothesis[i] = '?'

    return hypothesis



def display_hypothesis(hypothesis):
    print("\nFinal Hypothesis:")
    print(hypothesis)


def main():

    filename = "trainingdata.csv"

    data = load_data(filename)

    display_data(data)

    hypothesis = find_s_algorithm(data)

    display_hypothesis(hypothesis)


main()
```
