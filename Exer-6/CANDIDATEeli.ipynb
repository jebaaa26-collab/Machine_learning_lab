```python
import pandas as pd

def load_dataset(filename):
    data = pd.read_csv(filename)
    concepts = data.iloc[:, :-1].values
    target = data.iloc[:, -1].values
    return concepts, target



def display_dataset(concepts, target):
    print("Training Data:\n")
    for i in range(len(concepts)):
        print(list(concepts[i]), "->", target[i])


def initialize(concepts):
    specific = concepts[0].copy()
    general = [["?" for _ in range(len(specific))]]
    return specific, general



def candidate_elimination(concepts, target):


    specific, general = initialize(concepts)


    for i, instance in enumerate(concepts):

  
        if target[i].strip().lower() == "yes":

            for j in range(len(specific)):
                if specific[j] != instance[j]:
                    specific[j] = "?"

            general = [
                g for g in general
                if all(
                    g[k] == "?" or g[k] == specific[k]
                    for k in range(len(specific))
                )
            ]

        else:

            new_general = []

      
            for j in range(len(specific)):

                if instance[j] != specific[j]:

                    hypothesis = ["?" for _ in range(len(specific))]
                    hypothesis[j] = specific[j]

                    new_general.append(hypothesis)

      
            general.extend(new_general)

    return specific, general


def display_result(specific, general):

    print("\nFinal Specific Hypothesis (S):")
    print(list(specific))

    print("\nFinal General Hypotheses (G):")

    for g in general:
        print(g)


def main():

    filename = "trainingdata.csv"

    concepts, target = load_dataset(filename)

    display_dataset(concepts, target)

    specific, general = candidate_elimination(concepts, target)

    display_result(specific, general)


if __name__ == "__main__":
    main()
```
